```python
import py4hw

class NBitAdder(py4hw.Logic):
    def __init__(self, parent, name, a, b, r):
        super().__init__(parent, name)
        self.addIn('a', a)
        self.addIn('b', b)
        self.addOut('r', r)
        py4hw.Add(self, 'add', a, b, r)
```

```python
n = 8
hw = py4hw.HWSystem()
a = hw.wire('a', n)
b = hw.wire('b', n)
r = hw.wire('r', n)

py4hw.Constant(hw, 'ca', 25, a)
py4hw.Constant(hw, 'cb', 17, b)
adder = NBitAdder(hw, 'adder', a, b, r)

sch = py4hw.Schematic(adder)
sch.draw()
```
