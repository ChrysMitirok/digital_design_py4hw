```python
import py4hw
import py4hw.external.platforms.terasic as terasic
import py4hw.external.cygwin.edalize as cygwin
import py4hw.logic.protocol.uart as uart
```

```python
hw = terasic.DE10_Lite()
reset = hw.wire('reset')
tx_data, tx_ready, tx_valid = hw.getJTAGUART(reset)
py4hw.Constant(hw, 'reset', 0, reset)
uart.MsgSequencer(hw, 'sequencer', reset, tx_ready, tx_valid, tx_data, 'Hello world from py4hw\n\r')
```

```python
import time
t0 = time.time()
BASH = r"c:\cygwin64\bin\bash.exe"
QUARTUS_PATH = r"C:\altera_lite\25.1std\quartus\bin64"
project_dir = r"c:\temp\test_de10_lite"
cygwin.patch_edalize_for_quartus_project(project_dir, QUARTUS_PATH, BASH, verbose=True)
hw.build(project_dir)
tf = time.time()
print('BUILT OK', tf-t0, 'seconds')
py4hw.Sound.beep(duration=1)
```

```python
import time
t0 = time.time()
hw.download(project_dir)
tf = time.time()
print('DOWNLOAD OK', tf-t0, 'seconds')
py4hw.Sound.beep(duration=1)
```
